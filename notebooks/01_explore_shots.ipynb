```python
import sys
sys.path.append("../src")  # so we can import our own files

from load_data import load_shots
from shot_map import draw_shot_map

shots = load_shots()
shots.head()
```

Each row is one shot. `x` and `y` are where it was taken (the pitch is 120 x 80, the goal is at x = 120, y = 40). `is_goal` is what we want to predict.

```python
print(len(shots), "shots")
print(shots["is_goal"].sum(), "goals")
print(f"{shots['is_goal'].mean():.1%} of shots score")
```

Only about 1 shot in 9 scores (and that includes penalties). This is called **class imbalance**: a model that always says "no goal" would be right about 88% of the time and still be useless. That is why we will judge the model on its *probabilities*, not on accuracy.

```python
# How often does each kind of shot score?
shots.groupby("body_part")["is_goal"].agg(["count", "mean"])
```

```python
shots.groupby("shot_type")["is_goal"].agg(["count", "mean"])
```

```python
open_play = shots[shots["shot_type"] != "Penalty"]
draw_shot_map(open_play[open_play["competition"] == "World Cup 2022"], "Every non-penalty shot, World Cup 2022", "shot_map_world_cup_2022.png");
```
